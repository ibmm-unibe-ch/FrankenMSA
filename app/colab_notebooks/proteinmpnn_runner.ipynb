```python
# @title ProteinMPNN Colab Runner (FrankenMSA)
# @markdown ---
# This notebook runs ProteinMPNN directly on a Colab GPU and accepts parameters from the FrankenMSA web tool.

from google.colab import output
import urllib.parse as _u

def _parse_colab_url_params():
    search = output.eval_js("window.location.search") or ""
    hash_ = output.eval_js("window.location.hash") or ""
    qs = search[1:] if search.startswith("?") else ""
    if not qs and hash_.startswith("#?"):
        qs = hash_[2:]
    params = dict(_u.parse_qsl(qs))
    return params

PARAMS = _parse_colab_url_params()

print("Received parameters:")
print(PARAMS)
```
